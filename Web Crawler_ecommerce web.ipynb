```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

url = "https://books.toscrape.com/"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    products = soup.select("article.product_pod")

    for product in products:
        name_tag = product.select_one("h3 a")
        price_tag = product.select_one(".price_color")

        if name_tag and price_tag:
            name = name_tag.get("title", name_tag.get_text(strip=True))
            price = price_tag.get_text(strip=True)
            link = urljoin(url, name_tag.get("href", ""))

            print("Product Name:", name)
            print("Price:", price)
            print("Product Link:", link)
            print("-" * 40)

except requests.RequestException as e:
    print("Error accessing website:", e)
```
